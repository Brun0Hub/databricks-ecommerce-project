#Creating the Catalog and its Schemas - ecommerce_project

This notebook creates the foundational structure of the Databricks project using **Unity Catalog** (Lakeflow), including:

- Main Catalog: ecommnerce_project
- Schemas for the Medallion Architecture:
  - `00_landing` - Raw data landing zone
  - `01_bronze` - Raw data preserved in its original form
  - `02_silver` - Cleaned, validated, and enriched data
  - `03_gold` - Aggregated and business-ready data
  - `default` - Default schema
  - `information_schema` - System metadata

###1) Creating the Catalog (First level of the hierarqchy)

```sql
CREATE CATALOG IF NOT EXISTS ecommerce_catalog
COMMENT 'Main project Catalog - ecommerce_catalog'
```

```sql
USE CATALOG ecommerce_catalog;
SHOW SCHEMAS
```

###2) Creating the Schemas(Second Level of the hierarchy)

```sql
CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.00_landing
COMMENT 'Landing Zona - receives raw data from external systems'
```

```sql
CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.01_bronze
COMMENT 'Bronze layer - raw data preserved as is'
```

```sql
CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.02_silver
COMMENT 'Silver layer - Cleaned, validated, and enriched data'
```

```sql
CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.03_gold
COMMENT 'Gold Layer - Aggregated and business-ready data'
```

###3) Checking the object created for the project

```sql
SHOW CATALOGS LIKE 'ecommerce_catalog'
```

```sql
SHOW SCHEMAS IN ecommerce_catalog
```

```sql
DESCRIBE CATALOG ecommerce_catalog
```

```sql
DESCRIBE SCHEMA ecommerce_catalog.00_landing
```
