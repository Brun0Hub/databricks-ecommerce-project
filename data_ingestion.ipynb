```sql
CREATE TABLE IF NOT EXISTS ecommerce_catalog.`01_bronze`.products
AS SELECT *
FROM read_files(
    '/Volumes/ecommerce_catalog/00_landing/raw_files/products.csv',
    format => 'csv')
```

###Data Ingestion with COPY INTO

```sql
CREATE OR REPLACE TABLE ecommerce_catalog.`01_bronze`.orders(
    order_id STRING,
    customer_id STRING,
    order_date STRING,
    status STRING
)
```

```sql
COPY INTO ecommerce_catalog.`01_bronze`.orders
FROM '/Volumes/ecommerce_catalog/00_landing/raw_files/orders.csv'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header'='true')
COPY_OPTIONS ('mergeSchema' = 'false');
```

```sql
CREATE OR REPLACE TABLE ecommerce_catalog.`01_bronze`.order_items(
    order_item_id STRING,
    order_id STRING,
    product_id STRING,
    quantity STRING,
    unit_price STRING
)
```

```sql
COPY INTO ecommerce_catalog.`01_bronze`.order_items
FROM '/Volumes/ecommerce_catalog/00_landing/raw_files/order_items.csv'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header'='true')
COPY_OPTIONS ('mergeSchema' = 'false');
```
